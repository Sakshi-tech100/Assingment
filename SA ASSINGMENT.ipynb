````python
#1) Here’s how you can generate a list of 100 integers between 90 and 130 and perform the calculations in Python.

1. Generate the list of integers:
python
import random

int_list = [random.randint(90, 130) for _ in range(100)]


2. Calculate the mean:
python
def calculate_mean(numbers):
    return sum(numbers) / len(numbers)


3. Calculate the median:
python
def calculate_median(numbers):
    sorted_numbers = sorted(numbers)
    n = len(sorted_numbers)
    mid = n // 2
    if n % 2 == 0:
        return (sorted_numbers[mid - 1] + sorted_numbers[mid]) / 2
    else:
        return sorted_numbers[mid]


4. Compute the mode:
python
from collections import Counter

def calculate_mode(numbers):
    count = Counter(numbers)
    mode_data = count.most_common()
    max_count = mode_data[0][1]
    modes = [num for num, freq in mode_data if freq == max_count]
    return modes


5. Calculate the weighted mean:
python
def calculate_weighted_mean(values, weights):
    return sum(v * w for v, w in zip(values, weights)) / sum(weights)


6. Calculate the geometric mean:
python
import math

def calculate_geometric_mean(numbers):
    product = 1
    for number in numbers:
        product *= number
    return product ** (1 / len(numbers))
7. Calculate the harmonic mean:
python
def calculate_harmonic_mean(numbers):
    return len(numbers) / sum(1 / number for number in numbers)


8. Find the midrange:
python
def calculate_midrange(numbers):
    return (min(numbers) + max(numbers)) / 2


9. Calculate the trimmed mean:
python
def calculate_trimmed_mean(numbers, percentage):
    n = int(len(numbers) * percentage / 100)
    sorted_numbers = sorted(numbers)
    trimmed_numbers = sorted_numbers[n:-n]  # Remove the lowest and highest n elements
    return calculate_mean(trimmed_numbers)


Putting it all together, you can run these functions on your int_list to get the results for mean, median, mode, weighted mean, geometric mean, harmonic mean, midrange, and trimmed mean. 

For example, you can call these functions like this:
python
mean = calculate_mean(int_list)
median = calculate_median(int_list)
mode = calculate_mode(int_list)
# For weighted mean, you will need a weights list
weights = [1] * 100  # Example weights, replace as needed
weighted_mean = calculate_weighted_mean(int_list, weights)
geometric_mean = calculate_geometric_mean(int_list)
harmonic_mean = calculate_harmonic_mean(int_list)
midrange = calculate_midrange(int_list)
trimmed_mean = calculate_trimmed_mean(int_list, 10)  # Trimming 10% of outliers

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
print("Weighted Mean:", weighted_mean)
print("Geometric Mean:", geometric_mean)
print("Harmonic Mean:", harmonic_mean)
print("Midrange:", midrange)
print("Trimmed Mean:", trimmed_mean)
````

````python
#9 Create a program to calculate the cumulative distribution function (CDF) of exponential distribution
import numpy as np
import matplotlib.pyplot as plt
 
# Generate random discrete data
np.random.seed(0)
data = np.random.randint(0, 10, size=100)
 
# Calculate relative frequencies
values, counts = np.unique(data, return_counts=True)
probabilities = counts / counts.sum()
 
# Calculate the CDF
cdf = np.cumsum(probabilities)
 
# Plot the CDF
plt.step(values, cdf, where='post')
plt.xlabel('Value')
plt.ylabel('CDF')
plt.title('CDF of a Discrete Variable')
plt.grid(True)
plt.show()
````

```python

# 50 purchases out of 1000 visitors

# 70 purchases out of 1000 visitors

  ```

Apply z-test to find which layout is successful.

 For the old layout:
     - Number of purchases (successes) = 50
     - Total visitors = 1000
     - Conversion rate (p1) = 50 / 1000 = 0.05
   - For the new layout:
     - Number of purchases (successes) = 70
     - Total visitors = 1000
     - Conversion rate (p2) = 70 / 1000 = 0.07

2. Calculate the pooled proportion:
   The pooled proportion is calculated as follows:
   - Total successes = 50 + 70 = 120
   - Total visitors = 1000 + 1000 = 2000
   - Pooled proportion (p) = Total successes / Total visitors = 120 / 2000 = 0.06

3. Calculate the standard error (SE):
   The standard error for the difference in proportions can be calculated using the formula:
   SE = sqrt(p * (1 - p) * (1/n1 + 1/n2))
   where n1 and n2 are the sample sizes for the old and new layouts, respectively.
   - n1 = 1000 (old layout)
   - n2 = 1000 (new layout)
   - SE = sqrt(0.06 * (1 - 0.06) * (1/1000 + 1/1000))
   - SE = sqrt(0.06 * 0.94 * (0.001 + 0.001))
   - SE = sqrt(0.06 * 0.94 * 0.002) = sqrt(0.0001128) ≈ 0.0106

4. Calculate the z-score:
   The z-score can be calculated using the formula:
   z = (p1 - p2) / SE
   - z = (0.05 - 0.07) / 0.0106
   - z = -0.02 / 0.0106 ≈ -1.887

5. Determine the p-value:
   To find the p-value associated with the z-score, we can look it up in a standard normal distribution table or use a calculator. A z-score of -1.887 corresponds to a p-value of approximately 0.029.

6. Make a decision:
   Typically, a significance level (alpha) of 0.05 is used. Since the p-value (0.029) is less than 0.05, we reject the null hypothesis.

Conclusion: There is sufficient evidence to suggest that the new layout leads to a higher conversion rate compared to the old layout

````python
#14 1. State the Hypotheses:
   - Null Hypothesis (H0): The average response time is 5 minutes or more (μ ≥ 5).
   - Alternative Hypothesis (H1): The average response time is less than 5 minutes (μ < 5).

2. Collect Data:
   The response times are:
   4.3, 3.8, 5.1, 4.9, 4.7, 4.2, 5.2, 4.5, 4.6, 4.4.

3. Calculate the Sample Mean (x̄):
   To find the sample mean, sum all the response times and divide by the number of observations (n = 10).

   Sample Mean (x̄) = (4.3 + 3.8 + 5.1 + 4.9 + 4.7 + 4.2 + 5.2 + 4.5 + 4.6 + 4.4) / 10
   = 4.57 minutes.

4. Calculate the Sample Standard Deviation (s):
   First, find the variance by using the formula:
   Variance (s²) = Σ(xi - x̄)² / (n - 1)

   Calculate each (xi - x̄)²:
   - (4.3 - 4.57)² = 0.0729
   - (3.8 - 4.57)² = 0.5929
   - (5.1 - 4.57)² = 0.2809
   - (4.9 - 4.57)² = 0.1024
   - (4.7 - 4.57)² = 0.0169
   - (4.2 - 4.57)² = 0.1369
   - (5.2 - 4.57)² = 0.3969
   - (4.5 - 4.57)² = 0.0049
   - (4.6 - 4.57)² = 0.0009
   - (4.4 - 4.57)² = 0.0289

   Sum of squared differences = 0.0729 + 0.5929 + 0.2809 + 0.1024 + 0.0169 + 0.1369 + 0.3969 + 0.0049 + 0.0009 + 0.0289 = 1.542.

   Variance (s²) = 1.542 / (10 - 1) = 1.542 / 9 = 0.1713.

   Standard Deviation (s) = sqrt(0.1713) ≈ 0.414.

5. Calculate the z-score:
   The z-score is calculated using the formula:
   z = (x̄ - μ) / (s / sqrt(n))

   Here, μ = 5, n = 10, and we already calculated x̄ and s.

   z = (4.57 - 5) / (0.414 / sqrt(10))
   = (-0.43) / (0.414 / 3.1623)
   = (-0.43) / 0.1301
   ≈ -3.30.

6. Determine the Critical Value:
   For a one-tailed test at a significance level (alpha) of 0.05, the critical z-value is approximately -1.645.

7. Make a Decision:
   Since -3.30 is less than -1.645, we reject the null hypothesis.

8. Conclusion:
   There is sufficient evidence to support the claim that the average response time is less than 5 minutes. Therefore, the customer service department's claim is true.
````
