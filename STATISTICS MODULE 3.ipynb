Question2 : What is a p-value, and how is it used in hypothesis testing? What does it mean if the p-value is
very small (e.g., 0.01)

ans- A p-value measures the probability of obtaining the observed results, assuming that the null hypothesis is true. The lower the p-value, the greater the statistical significance of the observed difference. A p-value of 0.05 or lower is generally considered statistically significant.
